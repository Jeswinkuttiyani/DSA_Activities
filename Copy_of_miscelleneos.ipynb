```python
class employee:
    def __init__(self):
        self.name="Rahul"  # Public
        self._salary=5000  #protected
        self.__password="12345"  #private
    def show_details(self):
        print(self.name)
        print(self._salary)
        print(self.__password)

emp =employee()
print(emp.name)
print(emp._salary)
emp.show_details()
```

```python
class employee:
    def __init__(self):
        self.name="Rahul"  # Public
        self._salary=5000  #protected
        self.__password="12345"  #privat
    def show_parent(self):
        print("Name:",self.name)
        print("Salary:",self._salary)
        print("Password: ",self.__password)

class manager(employee):
  def show_manager(self):
    print("Name:",self.name)
    print("salary:",self._salary)
    # print("Password: ",self.__password)  not accessible
man=manager()
man.show_manager()
```
