```python
class Doctor:
    def __init__(self, doctor_id, doctor_name, specialization, available_slots):
        self.doctor_id = doctor_id
        self.doctor_name = doctor_name
        self.specialization = specialization
        self.available_slots = available_slots
        # {time: patient_name}
        self.appointments = {}


class AppointmentSystem:
    def __init__(self):
        self.doctors = {}

    # Book Appointment
    def book_appointment(self, patient_name, doctor_id, slot):
        # Check doctor
        if doctor_id not in self.doctors:
            print("Doctor Not Found")
            return

        doctor = self.doctors[doctor_id]
        # Check whether slots are available
        if doctor.available_slots == 0:
            print("Slot Not Available")
            return

        # Check whether patient already has appointment
        # with this doctor
        for booked_patient in doctor.appointments.values():
            if booked_patient == patient_name:
                print("Appointment Already Exists")
                return

        # Check whether this time is already booked
        if slot in doctor.appointments:
            print("Slot Not Available")
            return

        # Book appointment
        doctor.appointments[slot] = patient_name
        # Reduce available slots
        doctor.available_slots -= 1
        print("Appointment Booked")

    # Cancel Appointment
    def cancel_appointment(self, patient_name, doctor_id, slot):

        if doctor_id not in self.doctors:
            print("Appointment Not Found")
            return

        doctor = self.doctors[doctor_id]
        # Check appointment
        if (slot not in doctor.appointments or
                doctor.appointments[slot] != patient_name):
            print("Appointment Not Found")
            return

        # Remove appointment
        del doctor.appointments[slot]
        # Increase available slots
        doctor.available_slots += 1
        print("Appointment Cancelled")

    # Display Doctor Details
    def display_doctor(self, doctor_id):

        if doctor_id not in self.doctors:
            print("Doctor Not Found")
            return

        doctor = self.doctors[doctor_id]
        print(f"Doctor ID: {doctor.doctor_id}")
        print(f"Doctor Name: {doctor.doctor_name}")
        print(f"Specialization: {doctor.specialization}")
        print(f"Available Slots: {doctor.available_slots}")

    # Display Patient Appointments
    def display_patient(self, patient_name):

        found = False
        for doctor in self.doctors.values():
            for slot, patient in doctor.appointments.items():
                if patient == patient_name:
                    if not found:
                        print(f"\nAppointments for {patient_name}:")
                        found = True
                    print(f"{doctor.doctor_id} - {slot}")

        if not found:
            print("No Appointments")

    # Appointment Summary
    def summary(self):
        total_appointments = 0
        available_slots = 0
        most_booked_doctor = None
        highest_appointments = 0

        for doctor in self.doctors.values():
            appointment_count = len(doctor.appointments)
            total_appointments += appointment_count
            available_slots += doctor.available_slots

            if appointment_count > highest_appointments:
                highest_appointments = appointment_count
                most_booked_doctor = doctor.doctor_id

        print(f"\nTotal Appointments: {total_appointments}")

        if most_booked_doctor is not None:
            print(f"Most Booked Doctor: {most_booked_doctor}")
        else:
            print("Most Booked Doctor: None")

        print(f"Available Slots: {available_slots}")


# Main Program


system = AppointmentSystem()
# Number of doctors
n = int(input())
# Doctor details
for i in range(n):
    doctor_id, doctor_name, specialization, slots = input().split()
    doctor = Doctor(
        doctor_id,
        doctor_name,
        specialization,
        int(slots)
    )
    system.doctors[doctor_id] = doctor


# Number of operations
m = int(input())
for i in range(m):
    operation = input().split()
    command = operation[0]

    if command == "BOOK":
        patient_name = operation[1]
        doctor_id = operation[2]
        slot = operation[3]
        system.book_appointment(
            patient_name,
            doctor_id,
            slot
        )

    elif command == "CANCEL":
        patient_name = operation[1]
        doctor_id = operation[2]
        slot = operation[3]
        system.cancel_appointment(
            patient_name,
            doctor_id,
            slot
        )

    elif command == "DOCTOR":
        doctor_id = operation[1]
        system.display_doctor(doctor_id)

    elif command == "PATIENT":
        patient_name = operation[1]
        system.display_patient(patient_name)

    elif command == "SUMMARY":
        system.summary()
```
